```python
import math
import pandas as pd
```

```python
def simulate_pk_decay(initial_concentration, elimination_rate_k, total_hours):
    """
    Computes first-order exponential drug decay profiles hour-by-hour.
    Formula: C(t) = C0 * e^(-kt)
    """
    timeline_logs = []
    
    for hour in range(total_hours + 1):
        # Calculate continuous mathematical decay exponent
        current_concentration = initial_concentration * math.exp(-elimination_rate_k * hour)
        
        timeline_logs.append({
            'Hour': hour,
            'Concentration_mg_L': round(current_concentration, 3)
        })
        
    # Pipe results cleanly into a DataFrame for visualization
    return pd.DataFrame(timeline_logs)

# Verification execution test: 100 mg/L starting dose, k = 0.15 clearance rate
pk_dataframe = simulate_pk_decay(initial_concentration=100.0, elimination_rate_k=0.15, total_hours=24)
print(pk_dataframe.head(10))
```
